**K-Mens clustering **

```python
# Importing libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans, DBSCAN
from sklearn.decomposition import PCA
from sklearn.datasets import load_iris
```

```python
# Data loading
data = load_iris()
X = data.data
X
```

```python
inertia = []
k_values = range(1 , 11)

for k in k_values:
  kmeans = KMeans(n_clusters=k, random_state=42, n_init=10).fit(X)
  inertia.append(kmeans.inertia_)

print(inertia)

# Graphing
plt.figure(figsize=(8, 5))
plt.plot(k_values, inertia, marker = 'o')
plt.title('Elbow graph for K-Means algotithim')
plt.xlabel('Number of clusters')
plt.ylabel('Variance within Cluster (Inertia)')
plt.xticks(k_values)
plt.tight_layout()
plt.show()
```

```python
# Yuqoridagi tirsak graphidan ko'rdikki klasterlarimizning sonini 3 ta qilsak optimal bo'lar ekan
model = KMeans(n_clusters=3, random_state=42, n_init='auto').fit(X)
klas = model.labels_
klas
```

```python
# Let's make prediction
new_prediction = model.predict([[5.7, 3.2, 4.3, 1.7]])
new_prediction
```

**DBSCAN (Density Based Spatial Clustering of Application with Noise)**

*Core (doirani ichiga kirgan nuqtalar), Non-core (doirni ichiga kirmagan nuqtalar)*

```python
dbscan_model = DBSCAN(eps=0.3, min_samples=5).fit(X)
dbscan_model.labels_

#It seems that DBSCAN clustering could not cluster our dataset well compared to K-Means clustering
```

# PCA (Principal Component Analysis) or demensional reduction 3D-->2D

```python
pca_model = PCA(n_components=2).fit(X)
pca_model
```

```python
print(pca_model.explained_variance_ratio_)
print(pca_model.singular_values_)
```
